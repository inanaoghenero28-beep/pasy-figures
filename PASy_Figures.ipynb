# PASy: Figures 1–3

This notebook regenerates Figures 1–3 of the manuscript *Performance Modeling for Isolated Industrial Operations: A Data Envelopment Analysis Framework* (Inana, O.).

It runs `make_figures.py` on the two result files in `runs/`. No value is typed into the script: everything plotted is read from those files. The full data and results are in the Zenodo deposit cited in the manuscript.

**To use:** Runtime → Run all. It needs only the packages that Colab already has (numpy, pandas, matplotlib).

## 1. Get the script and the result files

```python
import os, subprocess, sys
REPO = "https://github.com/inanaoghenero28-beep/pasy-figures.git"
if not os.path.exists('make_figures.py'):
    if not os.path.isdir('pasy-figures'):
        subprocess.run(['git', 'clone', '--depth', '1', REPO], check=True)
    os.chdir('pasy-figures')
print('working in', os.getcwd())
print(sorted(os.listdir('runs')))
```

## 2. Generate the figures

```python
subprocess.run([sys.executable, 'make_figures.py', 'runs', 'figs'], check=True)
print(sorted(os.listdir('figs')))
```

## 3. Show the figures

```python
from IPython.display import Image, display
for n in (1, 2, 3):
    display(Image(filename=f'figs/Fig{n}.png'))
```

The figures are saved in `figs/` as EPS, PDF and PNG. In Colab, download them from the Files panel on the left.
